```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url='/kaggle/input/alzheimers-disease-patient-data/alzheimers_disease_patient_data.csv'
alzheimer = pd.read_csv(url)
alzheimer
```

```python
alzheimer.index
```

```python
alzheimer.isna().sum()
```

```python
alzheimer_describe = alzheimer.describe()
alzheimer_describe
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(alzheimer_describe.corr() , fmt='.2f'  )
```

```python
Age_describe = alzheimer['Age'].describe()
Age_describe
```

```python
rel_Diagnosis_Age = alzheimer.groupby('Age')['Diagnosis'].value_counts().unstack()
rel_Diagnosis_Age
```

```python
rel_Diagnosis_Age.plot(kind='bar' , stacked=True)
plt.xlabel('Age')
plt.ylabel('Diagnosis')
plt.title('Age Distribution by Diagnosis')
plt.show()
```
